```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-10,10,150)
plt.plot(x, 1/ (1+ np.exp(-x)), label= 'Sigmoid')
plt.title('Sigmoid Activaton Funtion')
plt.xlabel('Input')
plt.ylabel('Activation')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.plot(x, np.tanh(x), label= 'tanh', color='red')
plt.title('tanh Activaton Funtion')
plt.xlabel('Input')
plt.ylabel('Activation')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.plot(x, np.maximum(0,x), label= 'Relu', color='green')
plt.title('Maximum Activaton Funtion')
plt.xlabel('Input')
plt.ylabel('Activation')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.plot(x, x, label= 'Identity', color= 'orange')
plt.title('Identity Activaton Funtion')
plt.xlabel('Input')
plt.ylabel('Activation')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.plot(x, np.exp(x)/ np.sum(np.exp(x)), label= 'SoftMax' , color = 'purple')
plt.title('SoftMax Activaton Funtion')
plt.xlabel('Input')
plt.ylabel('Activation')
plt.legend()
plt.grid(True)
plt.show()
```
